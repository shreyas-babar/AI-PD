```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Name": ["Amit Sharma", "Neha Patil", "Rahul Verma", "Sneha Kulkarni"],
    "Email": ["amit@gmail.com", "neha@gmail.com", "rahul@gmail.com", "sneha@gmail.com"],
    "Phone": ["9876543210", "9123456780", "9988776655", "9871234567"],
    "Salary": [50000, 65000, 72000, 58000]
}

df = pd.DataFrame(data)

print("Original Dataset")
df
```

```python
sensitive_columns = ["Name", "Email", "Phone", "Salary"]

print("Sensitive Columns:")
print(sensitive_columns)
```

#### Mask Name

```python
def mask_name(name):
    return name[0] + "***"
```

#### Mask Email

```python
def mask_email(email):
    username, domain = email.split("@")
    return username[0] + "***@" + domain
```

#### Mask Phone

```python
def mask_phone(phone):
    return "XXXXX" + phone[-4:]
```

#### Shuffle Salaries

```python
def shuffle_salary(series):
    return np.random.permutation(series)
```

#### Apply Data Masking

```python
masked_df = df.copy()

masked_df["Name"] = masked_df["Name"].apply(mask_name)
masked_df["Email"] = masked_df["Email"].apply(mask_email)
masked_df["Phone"] = masked_df["Phone"].apply(mask_phone)
masked_df["Salary"] = shuffle_salary(masked_df["Salary"])
```

```python
print("Original Data")
display(df)

print("Masked Data")
display(masked_df)
```

```python
masked_df.to_csv("masked_dataset.csv", index=False)

print("Masked dataset saved successfully.")
```
