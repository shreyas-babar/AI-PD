```python
import pandas as pd
import numpy as np
import hashlib

df = pd.read_csv("masking_dataset_1000.csv")
```

```python
df
```

```python
def mask_name(name):
    return name[0]+"****"
```

```python
def mask_phone(phone):
    phone = str(phone)
    return "XXXXXX" + phone[-4:]
```

```python
def mask_email(email):
    username, domain = email.split("@")
    return username[0] + "***@" + domain
```

```python
import random

def mask_salary(salary_list):
    random.shuffle(salary_list)
    return salary_list
```

```python
masked_df = df.copy()
```

```python
masked_df["Email"] = df["Email"].apply(mask_email)
masked_df["Phone"] = df["Phone"].apply(mask_phone)
masked_df["Name"] = df["Name"].apply(mask_name)

# Salary needs list
masked_df["Salary"] = mask_salary(df["Salary"].tolist())
```

```python
print(df)
print(masked_df)
```

```python
masked_df.to_csv('making.csv',index=False)
```
