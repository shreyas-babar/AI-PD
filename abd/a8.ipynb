```python
df=pd.read_csv('OnlineRetail.csv',encoding='Latin-1')
```

```python
df = df[df['UnitPrice'] >= 0]
```

```python
df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

```python
df=df.groupby(['CustomerID','StockCode'])["Quantity"].sum().reset_index()
```

```python
df
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
user_item=df.pivot_table(columns='StockCode',
                    values='Quantity',
                    index='CustomerID').fillna(0)
```

```python
user_item
```

```python
similarity_item=cosine_similarity(user_item)
```

```python
similarity_df=pd.DataFrame(similarity_item,
                          columns=user_item.index,
                           index=user_item.index
                          )
```

```python
similarity_df
```

```python
from sklearn.cluster import KMeans
```

```python
kmeans = KMeans(n_clusters=5, random_state=42)
user_clusters = kmeans.fit_predict(user_item)
```

```python
user_item['Cluster'] = user_clusters

user_item['Cluster']
```

```python
user_item
```

```python
def recommend_products(user_id, top_n=5):
    user_id=int(user_id)

    similar_users = similarity_df.loc[user_id].sort_values(ascending=False)[1:3]

    similar_users_ids = similar_users.index

    recommended_products = df[df['CustomerID'].isin(similar_users_ids)]["StockCode"].value_counts()

    return recommended_products.head(top_n)
```

```python
recommend_products(similarity_df.index[0])
```
